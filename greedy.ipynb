```python
names = []
distances_matrix = []

def load_files(name_file, distance_file): 

    global distances_matrix
    global names

    with open(name_file, 'r') as NF:
                names = NF.readlines()
                for i in range(len(names)):
                    names[i] = names[i].strip()
                
                
    with open (distance_file, 'r') as DF:
            distances = DF.readlines()
            

            distances_matrix = []
            for nums_str in distances:
                distance_split = nums_str.split()
                for i in range(len(distance_split)):
                    distance_split[i] = float(distance_split[i])
                distances_matrix.append(distance_split)
```

```python
def nearest_city(remaining_cities, active_city):
        global distances_matrix 

        min_city = remaining_cities[0]
        min_distance = distances_matrix[active_city][min_city]

        #check every city in the list and if the distance is any closer 
        #then the minimum distance is updated
        for city in remaining_cities:
            if distances_matrix[active_city][city]< min_distance:
                min_distance = distances_matrix[active_city][city]
                min_city = city

        return min_city
     
```
