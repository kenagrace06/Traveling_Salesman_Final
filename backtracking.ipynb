This function should take as input the name and distance information found in name_file and distance_file. This function should only read in the files, put the data into a useful form (the distance information will need to be global), and perform the initial recursive call (described below). The recursion itself should be handled by a separate function (described below). The function tsp_backtrack should print the optimal tour (as a list of names of cities) as well as the total distance.

```python
names = []
distances_matrix = []

def load_files(name_file, distance_file): 

    global distances_matrix
    global names

    with open(name_file, 'r') as NF:
                names = NF.readlines()
                for i in range(len(names)):
                    names[i] = names[i].strip()
                
                
    with open (distance_file, 'r') as DF:
            distances = DF.readlines()
            

            distances_matrix = []
            for nums_str in distances:
                distance_split = nums_str.split()
                for i in range(len(distance_split)):
                    distance_split[i] = float(distance_split[i])
                distances_matrix.append(distance_split)
```

```python
distances_matrix = []   
best_solution = []
#have to set the distance to infinity
best_distance = float('inf')
names = []
start_city = 0

def tsp_recursive(partial_tour, remaining_cities):
    global best_solution
    global best_distance

    # if there are no cities left to visit, then the tour is done 
    if len(remaining_cities) == 0: 
        current_distance = 0 #start distance at 0
        length_tour = len(partial_tour)
        #for each city in the length of the tour so far 
        for city in range(length_tour):
            #current city a and next city b - uses % to loop back to the first city
            a = partial_tour[city]
            b = partial_tour[(city+1) % length_tour]
            #add distaace between a and b to current_distance
            current_distance = current_distance + distances_matrix[a][b]

        #if it found the shortest tour 
        if current_distance < best_distance:
                best_distance = current_distance #update best solution
                best_solution = partial_tour.copy()
        return

    #if there are more cities 
    for city in range(len(remaining_cities)):
          the_city = remaining_cities[city] # get city 
          #new list of remaining cities minus the one we are looking at right now
          next_remaining = remaining_cities[:city] + remaining_cities[city+1:]
          #current city is added to the tour
          partial_tour.append(the_city)

          #recursive !!
          tsp_recursive(partial_tour, next_remaining)

          #removes the city so that the loop will try a different city next time
          partial_tour.pop()
```

```python
def tsp_backtrack(name_file, distance_file):    
        global distances_matrix, best_solution, best_distance, names, start_city 

        load_files(name_file, distance_file)

        #list of all the cities without the first one (leaves out 0 by starting at 1)
        city_list = list(range(1, len(names)))

        for a in range(len(city_list)): 
            for b in range(a + 1, len(city_list)): 
                #city 1 is the second city in the tour and city 2 is the one to check
                city1 = city_list[a]
                city2 = city_list[b]

                #copy and remove 
                remaining_cities = city_list.copy()
                remaining_cities.remove(city1)
                remaining_cities.remove(city2)

                #starts tour to go 0, then city1 (second city)
                partial_tour = [start_city, city1]
                remaining = remaining_cities + [city2]

                tsp_recursive(partial_tour, remaining)

        final_tour = []
        #looks through the index of the tour to find the names of the cities and attatches them
        for city_index in best_solution:
              final_tour.append(names[city_index])


        print("Final tour:", final_tour)
        print("Total distance:", best_distance)
```
