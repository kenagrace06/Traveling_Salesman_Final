This function should take as input the name and distance information found in name_file and distance_file. It should print the TSP route (as a list of names of cities) generated by following the evolutionary procedure discussed below as well as the total distance.

```python
import random

#Returns distance
def calculation(distances_matrix, names, route): 
    total_distance = 0
    for i in range(len(route)-1):
        total_distance += distances_matrix[names.index(route[i])][names.index(route[i+1])]
    total_distance += distances_matrix[names.index(route[-1])][names.index(route[0])]
    return total_distance

#Given a TSP solution, make a small random change that is guaranteed to result in another solution.
def mutate_function(route):
    mutated_route = route.copy()
    random_city1 = random.randint(0,len(route)-1)
    random_city2 = random.randint(0,len(route)-1)

    while random_city1 == random_city2:
        random_city2 = random.randint(0,len(route)-1)

    mutated_route[random_city1], mutated_route[random_city2] = mutated_route[random_city2], mutated_route[random_city1]
    return mutated_route

names = []
distances_matrix = []

def load_files(name_file, distance_file): 

    global distances_matrix
    global names

    with open(name_file, 'r') as NF:
                names = NF.readlines()
                for i in range(len(names)):
                    names[i] = names[i].strip()
                
                
    with open (distance_file, 'r') as DF:
            distances = DF.readlines()
            

            distances_matrix = []
            for nums_str in distances:
                distance_split = nums_str.split()
                for i in range(len(distance_split)):
                    distance_split[i] = float(distance_split[i])
                distances_matrix.append(distance_split)

def tsp_evolution(name_file, distance_file):
    global names 
    global distances_matrix

    load_files(name_file, distance_file)

    #initialize
    current_parent = names.copy()
    current_parent_distance = calculation(distances_matrix, names, current_parent)

    #track best
    current_best_route = current_parent.copy()
    current_best_distance = current_parent_distance.copy()

    for generation in range(3):
        parent = current_best_route.copy()

        for i in range(10): 
            offspring = mutate_function(parent)
            offspring_distance = calculation(distances_matrix, names, offspring)

            if offspring_distance < best_distance:
                current_best_route = offspring.copy()
                best_distance = offspring_distance

        print("generation: ", generation+1, "Weight: ", best_distance, current_best_route)
```
